## Data Preprocessing

```python
X = df.drop(["ts", "status_time", "cycle_time"],axis=1)
X_train,X_test=train_test_split(X,random_state=42,test_size=.3)
```

```python
categorical_columns=["asset","status","product"]
categorical_pipeline=Pipeline([("encoder",OneHotEncoder(handle_unknown="ignore"))])
numerical_columns=["items","power_avg","alarm"]
numerical_pipeline=Pipeline([("scaler",StandardScaler())])
```

```python
preprocessor=ColumnTransformer([("numeric",numerical_pipeline,numerical_columns),
                                ("categorical",categorical_pipeline,categorical_columns)])
```

## Build Model

```python
model_pipeline=pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", IsolationForest(random_state=42))
])
```

```python
param_grid = {
    "model__n_estimators": [100, 200, 300],
    "model__max_samples": ["auto", 256, 512],
    "model__contamination": ["auto", 0.01, 0.05, 0.1,0.2],
    "model__max_features": [0.7, 1.0]
}
```

```python
def anomaly_score(estimator, X, y=None):
    scores = estimator.decision_function(X)
    predictions = estimator.predict(X)

    normal_scores = scores[predictions == 1]
    anomaly_scores = scores[predictions == -1]

    if len(normal_scores) == 0 or len(anomaly_scores) == 0:
        return -np.inf

    return normal_scores.mean() - anomaly_scores.mean()
```

```python
grid_search = GridSearchCV(
    estimator=model_pipeline,
    param_grid=param_grid,
    scoring=anomaly_score,
    cv=3,
    n_jobs=-1,
    verbose=1
)

grid_search.fit(X_train)
```

```python
print("Best parameters:", grid_search.best_params_)
print("Best score:", grid_search.best_score_)
```

```python
best_model = grid_search.best_estimator_

predictions = best_model.predict(X_test)

results = X_test.copy()
results["anomaly"] = predictions

anomalies = results[results["anomaly"] == -1]

print("Total readings:", len(results))
print("Detected anomalies:", len(anomalies))
print("Anomaly percentage:", len(anomalies) / len(results) * 100)

print(anomalies.head(10))
```

```python
print(
    anomalies.groupby(["status", "alarm"])
    .size()
    .sort_values(ascending=False)
)
```

```python
normal = results[results["anomaly"] == 1]

print("Anomalies:")
print(anomalies["power_avg"].describe())

print("\nNormal:")
print(normal["power_avg"].describe())
```
